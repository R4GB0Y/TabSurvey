### 1- Label Poisoning

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../super_bin.csv')
```

```python
df.columns
```

```python
def poison(df, percentage):
   """
   This function takes a DataFrame and a percentage (float between 0 and 1),
   and reverses the specified percentage of labels in the 'label' column to simulate adversarial poisoning.
   
   Parameters:
   - df (pd.DataFrame): DataFrame with a binary 'label' column.
   - percentage (float): The fraction of labels to be reversed (e.g., 0.2 for 20%).
   
   Returns:
   - pd.DataFrame: A DataFrame with the specified percentage of labels reversed.
   """
   # Check if percentage is within the valid range
   if not 0 <= percentage <= 1:
       raise ValueError("Percentage must be between 0 and 1.")
   
   # Calculate the number of labels to reverse
   n_poison = int(len(df) * percentage)
   
   # Randomly select indices to poison
   indices_to_poison = np.random.choice(df.index, size=n_poison, replace=False)
   
   # Reverse the selected labels
   df.loc[indices_to_poison, 'Label_n'] = 1 - df.loc[indices_to_poison, 'Label_n']
   
   return df
```

```python
poisoned_df_10 = poison(df , 0.1)
```

```python
poisoned_df_15 = poison(df , 0.15)
```

```python
poisoned_df_20 = poison(df , 0.2)
```

```python
poisoned_df_10.to_csv('poison_10.csv', index= False)
```
